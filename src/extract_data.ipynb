```python
# src/make_clips.py
import numpy as np, pandas as pd, librosa, librosa.display, matplotlib.pyplot as plt

labels = pd.read_csv("data/processed/labels.csv").reset_index(drop=True)
instruments = labels["instrument"].unique()
clips = np.zeros((len(labels), 2 * 16000), dtype=np.float16)

cached_audio = {}
for row_index, label in labels.iterrows():
        if "data/audio/" + label.source_file not in cached_audio:
            y, _ = librosa.load("data/audio/" + label.source_file, sr=16000, mono=True)
            cached_audio["data/audio/" + label.source_file] = y
        else:
            y = cached_audio["data/audio/" + label.source_file]
        a = int(label.start_s * 16000)
        seg = y[a : a + 2 * 16000]
        clips[row_index, : len(seg)] = seg

np.save("data/processed/clips_16k.npy", clips)
```
